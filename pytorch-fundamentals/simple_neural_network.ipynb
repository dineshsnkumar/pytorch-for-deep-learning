```python
import torch
import torch.nn as nn
import torch.optim as optim
```

### ML Pipeline

### 1. Data Ingestion and Preperation

```python
distances = torch.tensor([
    [1.0], [1.5], [2.0], [2.5], [3.0], [3.5], [4.0], [4.5], [5.0], [5.5],
    [6.0], [6.5], [7.0], [7.5], [8.0], [8.5], [9.0], [9.5], [10.0], [10.5],
    [11.0], [11.5], [12.0], [12.5], [13.0], [13.5], [14.0], [14.5], [15.0], [15.5],
    [16.0], [16.5], [17.0], [17.5], [18.0], [18.5], [19.0], [19.5], [20.0]
], dtype=torch.float32)

times = torch.tensor([
    [6.96], [9.67], [12.11], [14.56], [16.77], [21.7], [26.52], [32.47], [37.15], [42.35],
    [46.1], [52.98], [57.76], [61.29], [66.15], [67.63], [69.45], [71.57], [72.8], [73.88],
    [76.34], [76.38], [78.34], [80.07], [81.86], [84.45], [83.98], [86.55], [88.33], [86.83],
    [89.24], [88.11], [88.16], [91.77], [92.27], [92.13], [90.73], [90.39], [92.98]
], dtype=torch.float32)
```

#### 1.1 Data Normalization

Data normalizaton is used to make the training process stable and prevent the large distances value from dominating the learning process. It also keeps the gradient stable during training

```python
distances_mean = distances.mean()
distances_std = distances.std()

times_mean = times.mean()
times_std = times.std()

distances_norm = (distances- distances_mean)/ (distances_std)
times_norm = (times - times_mean)/ (times_std)
```

#### 2. Model Building

```python
torch.manual_seed(42)


model = nn.Sequential(
    nn.Linear(1, 3),
    nn.ReLU(),
    nn.Linear(3,1)
)
```

The goal is to find best weights and bias that fit the non-linear equation which has a single input (*distances*) and the output (times). For this PyTorch provides a `nn.Sequential()`.

`nn.Linear(1,3)` means that it is taking one input(distances) and producing three neurons. It is the hidden layer.

`nn.ReLU()` is the activation function to the output of the three neurons.

`nn.Linear(3, 1)` means that we take the output of the three neurons and produce single output(time)


#### 3. Training

```python
loss_function = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

`nn.MSELoss()` is the Mean Squared Error loss function.

`optim.SGD()` is the Stochastic Gradient Descent optimizer. Which adjusts the weights and bias based on the error.

`lr` is the learning rate. It adjusts how big each adjustment should be

```python
for epoch in range(3000):
  # Clear the gradients
  optimizer.zero_grad()
  # Make predictions (Forward pass)
  outputs = model(distances_norm)
  # Calculate the loss
  loss = loss_function(outputs, times_norm)
  # Calculate adjustments (Backward pass)
  loss.backward()
  # Update the model parameters
  optimizer.step()

  if (epoch + 1) % 50 == 0:
    print(f"Epoch {epoch+1} : Loss {loss.item()}")
```

#### Making predictions

```python
distance_to_predict = 7.0

with torch.no_grad():
  new_distance_tensor = torch.tensor([distance_to_predict], dtype=torch.float32)
  new_distance_norm = (new_distance_tensor - distances_mean)/ (distances_std)
  # Pass. the distance to predict to the model
  prediction_norm = model(new_distance_norm)
  prediction_actual_time = (prediction_norm * times_std) + times_mean
  print(f"The predicted time to travel {distance_to_predict} is {prediction_actual_time.item():.1f}")
```
